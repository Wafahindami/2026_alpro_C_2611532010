## 2.1 Dasar Teori
Dalam praktikum modul pertama ini, terdapat tiga komponen utama yang dipelajari:

1. **Tipe Data Numerik:** Python mendukung pengolahan data bilangan bulat (integer/`int`), desimal pecahan (`float`), dan bilangan kompleks (`complex`) untuk kebutuhan algoritma kalkulasi aritmatika.
2. **Tipe Data String:** Digunakan untuk mendeklarasikan objek berbentuk teks. Dapat diapit kutip satu (`'`), dua (`"`), ataupun tiga (`"""`) apabila berisi banyak baris (multiline).
3. **Tipe Data Boolean:** Sebuah tipe data esensial yang hanya dapat menyimpan nilai `True` atau `False`, yang sering kali dimanfaatkan pada pembuatan cabang percabangan logika program.
4. **Input Data Dinamis:**  Berbeda dengan nilai statis yang ditentukan langsung di dalam kode, instruksi input() memungkinkan sistem untuk meminta dan menangkap masukan secara langsung dari pengguna saat program dijalankan. 
5. **Konstanta:** Meski Python tidak mengekang deklarasi data, konstanta lazim digunakan untuk nilai yang disepakati mutlak dan dilarang berubah di tengah jalan, contohnya melalui bantuan dekorator `Final`.

### 2.2.2 Program "Numerik_2611532010"
Program berikut mengilustrasikan penciptaan beberapa variabel, masing-masing dengan muatan tipe numerik yang berbeda (integer, float, complex), kemudian mengidentifikasi validasi struktur bawaannya memakai function type()

```python
#buat file dengan nama Numerik_NIM.py
#nama variabel ditambah 4 digit terakhir NIM contoh : a_5500
#create a variable with numeric type value
a_2010 = 100
print("the type of variable having value", a_2010, "is", type(a_2010))

#create a variable with float value
b_2010=10.2010
print("the type of variable having value", b_2010, "is", type(b_2010))

#create a variable with complex value
c_2010 =100+3j
print ("the type of variable having value", c_2010, "is", type (c_2010))
```

### 2.2.3 Program "Konstanta_2611532010"
Menggunakan import atribut `Final` dari modul `typing`, kode ini mempatenkan variabel global `PI` senilai 3.14. Tujuannya adalah mereduksi kesalahan perubahan nilai sembarang secara tidak sengaja ketika program dikomputasi untuk menelusuri luas lingkaran.

```python
# Buat file dengan nama Konstanta_NIM.py
# Program ini menggunakan konstanta untuk menghitung luas lingkaran
# nama variabel ditambah 4 digit nim terakhir contoh: jari_1234

from typing import Final
PI: Final = 3.14
print ("pi: %f" % (PI))
jari_2010 = float(input("masukkan nilai jari jari: "))
luas_2010 = PI * jari_2010 * jari_2010
print("Luas lingkaran dengan jari jari %.2f adalah %.2f" % (jari_2010, luas_2010))
```

Penjelasan program string di atas:

1. `from typing import Final` : Berfungsi untuk memanggil library atau modul bawaan `typing` dari Python, guna menggunakan kemampuan pelabelan (anotasi) jenis variabel tertentu dalam hal ini sebagai tipe `Final`.
2. `PI: Final = 3.14` : Mendeklarasikan bahwa variabel `PI` bersifat final atau konstan. Ini menunjukkan pada struktur kode bahwa `PI` merupakan tetapan nilai matematis yang tidak boleh diubah oleh baris kode mana pun secara tidak sengaja di masa mendatang.
3. `float(input(...))` : Meminta input dari pengguna, namun berkat pembungkusan fungsi tipe data `float(...)`, teks angka yang dimasukkan pengguna secara sekejap langsung dikonversi ke tipe bilangan desimal (float).
4. `%f` dan `%.2f` : Berfungsi sebagai teknik pemformatan string kuno gaya bahasa C (*String Formatting*). Simbol `%f` mencetak bilangan float secara utuh, sedangkan `%.2f` digunakan untuk membatasi tampilan hasil hitung luas dan input jari-jari agar hanya menyisakan dua digit angka spesifik saja setelah tanda koma.

### 2.2.4 Program "Input_2611532010"
Implementasi fungsi `input()` pada program ini bertujuan agar variabel masukan bersifat fleksibel dari sisi pengguna. Data masukan tersebut awalnya terekam sebagai string, lalu diubah secara paksa `(type casting)` menjadi tipe `int` demi kelancaran kalkulasi dimensi.

```python
# Buat file dengan nama INput_NIM.py
# Buat program untuk menghitung luas persegi panjang
# Nama variabel ditambah 4 digit nim terakhir contoh: panjang_1234
# Program ini menggunakan fungsi input() untuk meminta pengguna
# Memasukkan nilai panjang dan lebar persegi panjang
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
Panjang_2010 = input('masukkan nilai panjang: ')
Lebar_2010 = input('masukkan nilai lebar: ')
Luas_2010 = int(Panjang_2010) * int(Lebar_2010)
print("luas =", Luas_2010)
```

Penjelasan program Input di atas:

1. `input('Masukkan nilai panjang: ')` : Berfungsi menampilkan teks sapaan ke layar, kemudian menahan eksekusi program sejenak untuk menunggu pengguna mengetikkan sesuatu (inputan) ke dalam *console* / *terminal*. Secara bawaan (default), semua masukan akan direkam sebagai data String.
2. `int(panjang_2010)` dan `int(lebar_2010)` : Berfungsi melakukan proses *type casting* atau pemaksaan konversi tipe data.[cite: 2] Nilai panjang dan lebar yang awalnya berupa karakter teks atau String diubah paksa menjadi bilangan bulat (Integer) agar dapat dikalikan dalam operasi matematika `*` (perkalian).
3. `luas_2010 = ...` : Menyimpan hasil akhir nilai perkalian yang bertipe Integer.

### 2.2.5 Program "Boolean_2611532010"
Kode program berikut mengimplementasikan variabel boolean untuk memvalidasi pemenuhan syarat kelulusan seorang mahasiswa. Status pencapaian nilai akan dievaluasi kebenarannya (`True` atau `False`) terlebih dahulu sebelum digunakan untuk memicu pencetakan pernyataan predikat kelulusan Cumlaude.

```python
#buat dile dengan nama Boolean_NIM.py
#nama variabel ditambah 4 digit terakhir NIM contoh : Nilai_1234
#Deklarasi variabel dengan tipe data boolean
is_lulus_2010 = True
is_cumlaude_2010 = True

# menggunakan Boolean
Nilai_2010 = 85
batas_lulus_2010 = 75

# Menentukan nilai boolean dari kondisi
status_kelulusan = Nilai_2010 >= batas_lulus_2010 #hasilnya akan True

print("=== check kelulusan ===")
print("Nilai_2010?:", status_kelulusan)
if is_lulus_2010 and is_cumlaude_2010:
    print("selamat, anda lulus dengan predikat Cumlaude!")
```

2. `status_kelulusan_2010 = nilai_2010 >= batas_lulus_2010` : Berfungsi melakukan evaluasi operator perbandingan relasional (lebih besar sama dengan). Karena nilai 85 lebih besar dari 75, hasil evaluasi tersebut akan mengembalikan nilai kebenaran Boolean berupa `True` lalu disalurkan ke variabel `status_kelulusan_2010`.
