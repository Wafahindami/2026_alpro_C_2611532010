## 2.1 Dasar Teori
Dalam praktikum modul ketiga ini, dibahas enam kelompok operator utama dalam Python:

1. **Operator Aritmatika:** Digunakan untuk melakukan operasi matematika dasar seperti penjumlahan (+), pengurangan (-), perkalian (*), pembagian (/), pembagian bulat (//), modulus (%), dan perpangkatan (**).
2. **Operator Perbandingan:** Digunakan untuk membandingkan dua nilai dan menghasilkan nilai kebenaran Boolean (`True` atau `False`), meliputi `>`, `<`, `>=`, `<=`, `==`, dan `!=`.
3. **Operator Logika:** Digunakan untuk menghubungkan ekspresi relasional atau boolean, terdiri dari `and`, `or`, dan `not`.
4. **Operator Penugasan (Assignment):** Digunakan untuk memberikan nilai ke dalam variabel secara ringkas, seperti `=`, `+=`, `-=`, `*=`, `/=`, dan lain-masing variasinya.
5. **Operator Keanggotaan (Membership):** Digunakan untuk menguji apakah suatu nilai terdapat dalam tipe data koleksi (seperti list atau tuple) menggunakan kata kunci `in` dan `not in`.
6. **Operator Identitas:** Digunakan untuk membandingkan lokasi memori dari dua objek menggunakan kata kunci `is` dan `is not`.
7. **Operator Bitwise:** Bekerja secara langsung pada representasi level bit (biner) dari bilangan bulat, meliputi `&` (AND), `|` (OR), `^` (XOR), `~` (NOT), `<<` (Shift Left), dan `>>` (Shift Right).

## 2.2 Kode Program

### 2.2.1 Program "Aritmatika_2611532010"
Program ini mempraktikkan penggunaan operator kalkulasi matematika dasar berdasarkan masukan angka dinamis dari pengguna yang dikonversi ke tipe integer.

```python
# buat file dengan nama aritmatika_2611532010.py
angka1_2010 = int(input("Input angka-1: "))
angka2_2010 = int(input("Input angka-2: "))

# Penjumlahan
hasil_2010 = angka1_2010 + angka2_2010
print("\nOperator Penjumlahan")
print("hasil =", hasil_2010)

# Pengurangan
hasil_2010 = angka1_2010 - angka2_2010
print("\nOperator Pengurangan")
print("hasil =", hasil_2010)

# Perkalian
hasil_2010 = angka1_2010 * angka2_2010
print("\nOperator Perkalian")
print("hasil =", hasil_2010)

# Pembagian, pembagian bulat, dan sisa bagi
if angka2_2010 != 0:
    hasil_2010 = angka1_2010 / angka2_2010
    print("\nOperator Pembagian")
    print("hasil =", hasil_2010)

    hasil_2010 = angka1_2010 // angka2_2010
    print("\nOperator Pembagian bulat")
    print("hasil =", hasil_2010)

    hasil_2010 = angka1_2010 % angka2_2010
    print("\nOperator Sisa bagi")
    print("hasil =", hasil_2010)
else:
    print("angka kedua tidak boleh bernilai 0,")

# Pangkat
hasil_2010 = angka1_2010 ** angka2_2010
print("\nOperator Pangkat")
print("hasil =", hasil_2010)
```

### 2.2.2 Program "Perbandingan_2611532010"
Program ini membandingkan hubungan nilai numerik antar variabel dan mencetak hasil evaluasi berupa Boolean.

```python
# Buat file dengan nama perbandingan_2611532010.py
angka1_2010 = int(input("Input angka-1: "))
angka2_2010 = int(input("Input angka-2: "))

# Lebih besar dari 
hasil_2010 = angka1_2010 > angka2_2010
print("\nOperator Lebih besar dari")
print("angka1 > angka2 =", hasil_2010)

# Lebih kecil dari
hasil_2010 = angka1_2010 < angka2_2010
print("\nOperator Lebih kecil dari")
print("angka1 < angka2 =", hasil_2010)

# lebih besar dari atau sama dengan
hasil_2010 = angka1_2010 >= angka2_2010
print("\nOperator Lebih besar dari atau sama dengan")
print("angka1 >= angka2 =", hasil_2010)

# lebih kecil dari atau sama dengan
hasil_2010 = angka1_2010 <= angka2_2010
print("\nOperator lebih kecil dari atau sama dengan")
print("angka1 <= angka2 =", hasil_2010)

# sama dengan
hasil_2010 = angka1_2010 == angka2_2010
print("\nOperator sama dengan")
print("angka1 == angka2 =", hasil_2010)

# tidak sama dengan
hasil_2010 = angka1_2010 != angka2_2010
print("\nOperator tidak sama dengan")
print("angka1 != angka2 =", hasil_2010)

# Tambahan: perbandingan berantai dalam python
hasil_2010 = 0 < angka1_2010 < 100
print("\nPerbandingan berantai")
print("0 < angka1 < 100 =", hasil_2010)

hasil_2010 = 0 < angka2_2010 < 100
print("0 < angka2 < 100 =", hasil_2010)
```

### 2.2.3 Program "Logika_2611532010"
Program ini mengevaluasi berbagai gerbang logika menggunakan variabel bertipe Boolean hasil dari konversi String masukan pengguna.

```python
# Buat file dengan nama logika_2611532010.py
a1_2010 = input("input nilai boolean-1 (True/False): ").strip().lower() == "true"
a2_2010 = input("input nilai boolean-2 (True/False): ").strip().lower() == "true"

print("\nA1 =", a1_2010)
print("A2 =", a2_2010)

# Konjungsi
hasil_2010 = a1_2010 and a2_2010
print("\nkonjungsi (and)")
print("A1 and A2 =", hasil_2010)

# Disjungsi
hasil_2010 = a1_2010 or a2_2010
print("\nDisjungsi (OR)")
print("A1 or A2 =", hasil_2010)

# Negasi A1
hasil_2010 = not a1_2010
print("\nNegasi A1 (NOT)")
print("not A1 =", hasil_2010)

# Negasi A2
hasil_2010 = not a2_2010
print("\nNegasi A2 (NOT)")
print("not A2 =", hasil_2010)

# XOR
hasil_2010 = a1_2010 != a2_2010
print("\nDisjungsi Eksklusif (XOR)")
print("A1 XOR A2 =", hasil_2010)
```

### 2.2.4 Program "Assignment_2611532010"
Program ini mendemonstrasikan penyederhanaan penulisan kode melalui pemakaian *compound assignment operators*.

```python
# Buat file dengan nama assigntment_2611532010.py
angka1_2010 = int(input("Input angka-1: "))
angka2_2010 = int(input("Input angka-2: "))

print("\nNilai awal angka1 =", angka1_2010)
print("Nilai awal angka2 =", angka2_2010)

# Assignment biasa
hasil_2010 = angka1_2010
print("\nOperator Assignment Biasa (=)")
print("hasil =", hasil_2010)

# Assignment penambahan 
hasil_2010 = angka1_2010
hasil_2010 += angka2_2010
print("\nAssignment penambahan (+=)")
print("Hasil =", hasil_2010)

# Assignment pengurangan
hasil_2010 = angka1_2010
hasil_2010 -= angka2_2010
print("\nAssignment pengurangan (-=)")
print("hasil =", hasil_2010)

# Assignment perkalian
hasil_2010 = angka1_2010
hasil_2010 *= angka2_2010
print("\nAssignment perkalian (*=)")
print("hasil =", hasil_2010)

# Assignment pembagian, pembagian bulat, dan sisa bagi
if angka2_2010 != 0:
    hasil_2010 = angka1_2010
    hasil_2010 /= angka2_2010
    print("\nAssignment pembagian (/=)")
    print("hasil =", hasil_2010)
    
    hasil_2010 = angka1_2010
    hasil_2010 //= angka2_2010
    print("\nAssignment pembagian bulat (//=)")
    print("Hasil =", hasil_2010)
    
    hasil_2010 = angka1_2010
    hasil_2010 %= angka2_2010
    print("\nAssignment sisa bagi (%=)")
    print("Hasil =", hasil_2010)
else:
    print("\nPembagian tidak dapat dilakukan.")

# Assignment perpangkatan
hasil_2010 = angka1_2010
hasil_2010 **= angka2_2010
print("\nAssignment perpangkatan (**=)")
print("Hasil =", hasil_2010)
```

### 2.2.5 Program "Bitwise_2611532010"
Program ini mendemonstrasikan manipulasi data desimal bulat langsung pada representasi level untaian bit biner.

```python
# Buat file dengan nama bitwise_2611532010.py
print("========================================")
print("3. OPERATOR BITWISE")
print("========================================")

angka1_2010 = int(input("Masukkan angka bitwise-1: ")) 
angka2_2010 = int(input("Masukkan angka bitwise-2: "))

print("\nAngka dalam bentuk desimal dan biner")
print("angka1 =", angka1_2010, "| biner =", bin(angka1_2010))
print("angka2 =", angka2_2010, "| biner =", bin(angka2_2010))

# Bitwise AND
hasil_2010 = angka1_2010 & angka2_2010
print("\nBitwise AND (&)")
print(angka1_2010, "&", angka2_2010, "=", hasil_2010)
print("Biner hasil =", bin(hasil_2010))
print("Biner hasil (8 bit) =", format(hasil_2010, "08b"))

# Bitwise OR
hasil_2010 = angka1_2010 | angka2_2010
print("\nBitwise OR (|)")
print(angka1_2010, "|", angka2_2010, "=", hasil_2010)
print("Biner hasil =", bin(hasil_2010))
print("Biner hasil (8 bit) =", format(hasil_2010, "08b"))

# Bitwise XOR
hasil_2010 = angka1_2010 ^ angka2_2010
print("\nBitwise XOR (^)")
print(angka1_2010, "^", angka2_2010, "=", hasil_2010)
print("Biner hasil =", bin(hasil_2010))
print("Biner hasil (8 bit) =", format(hasil_2010, "08b"))

# Bitwise NOT
hasil_2010 = ~angka1_2010
print("\nBitwise NOT (~)")
print("~", angka1_2010, "=", hasil_2010)
print("Biner hasil =", bin(hasil_2010))
print("Biner hasil (8 bit) =", format(hasil_2010, "08b"))

# Bitwise geser kiri
jumlah_geser_2010 = int(input("\nMasukkan jumlah pergeseran bit: "))

hasil_2010 = angka1_2010 << jumlah_geser_2010
print("\nBitwise geser kiri (<<)")
print(angka1_2010, "<<", jumlah_geser_2010, "=", hasil_2010)
print("Biner hasil =", bin(hasil_2010))
print("Biner hasil (8 bit) =", format(hasil_2010, "08b"))

# Bitwise geser kanan
hasil_2010 = angka1_2010 >> jumlah_geser_2010
print("\nBitwise geser kanan (>>)")
print(angka1_2010, ">>", jumlah_geser_2010, "=", hasil_2010)
print("Biner hasil =", bin(hasil_2010))
print("Biner hasil (8 bit) =", format(hasil_2010, "08b"))
```

### 2.2.6 Program "Lainnya_2611532010"
Program ini memvalidasi keanggotaan elemen data dalam suatu koleksi serta membandingkan kesamaan referensi lokasi alokasi memori fisik antar objek.

```python
# Buat file dengan nama lainnya_2611532010.py
print("========================================")
print("1. OPERATOR KEANGGOTAAN")
print("========================================")

input_data_2010 = input("Masukkan beberapa angka, pisahkan dengan koma: ")
data_2010 = [int(angka.strip()) for angka in input_data_2010.split(",")]
nilai_dicari = int(input("Masukkan angka yang ingin dicari: "))

# Operator in
hasil_2010  = nilai_dicari in data_2010
print("\nOperator keanggotaan IN")
print(nilai_dicari, "in", data_2010, "=", hasil_2010)

# Operator not in
hasil_2010 = nilai_dicari not in data_2010
print("\nOperator keanggotaan NOT IN")
print(nilai_dicari, "not in", data_2010, "=", hasil_2010)

print("\n========================================")
print("2. OPERATOR IDENTITAS")
print("========================================")

objek1_2010 = data_2010
objek2_2010 = objek1_2010
objek3_2010 = data_2010.copy()

print("objek1 =", objek1_2010)
print("objek2 =", objek2_2010)
print("objek3 =", objek3_2010)

# Operator is
hasil_2010 = objek1_2010 is objek2_2010
print("\nOperator identitas IS")
print("objek1 is objek2 =", hasil_2010)

# Operator is not
hasil_2010 = objek1_2010 is not objek3_2010
print("\nOperator identitas IS NOT")
print("objek1 is not objek3 =", hasil_2010)

# Membandingkan identitas dan nilai
print("\nPerbandingan identitas dan nilai")
print("objek1 is objek3 =", objek1_2010 is objek3_2010)
print("objek1 == objek3 =", objek1_2010 == objek3_2010)
```
