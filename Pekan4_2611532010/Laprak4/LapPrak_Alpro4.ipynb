## 2.1 Dasar Teori
Conditional merupakan struktur dalam Python yang digunakan untuk menjalankan perintah berdasarkan kondisi tertentu. Dengan conditional, program dapat menentukan apakah suatu perintah akan dijalankan atau tidak berdasarkan hasil evaluasi kondisi.

1. Pernyataan if
Pernyataan if digunakan untuk menjalankan suatu perintah apabila kondisi yang diberikan bernilai benar (True).
umur = 18
2. Pernyataan if-else
Pernyataan if-else digunakan ketika terdapat dua kemungkinan kondisi. Jika kondisi pada if bernilai benar, maka perintah pada if dijalankan. Jika salah, program menjalankan perintah pada else.
3. Pernyataan if-elif-else
if-elif-else digunakan ketika terdapat lebih dari dua kondisi yang perlu diperiksa. Python akan memeriksa kondisi secara berurutan hingga menemukan kondisi yang bernilai benar.

### 2.2.4 Program multi_if2_2611532010
</div>

```python
# Buat file dengan nama multi_if2_nim.py
# Buat program untuk kodisional if
# Nama variabel ditambah 4 digit terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input()
# Program Menghitung Diskon Belanja

# Input dan user
total_belanja_2010 = float(input("Input Total Belanja (Rp): "))

# Input status member (mengecek apakah user mengetik 'y atau 't')
input_member_2010 = input("Apakah Anda Member (y/t): ").strip().lower()
is_member_2010 = input_member_2010 in ["y", "t"]

# Input satus kode promo (mengecek apakah user mengetik 'y atau 't')
input_promo_2010 = input("Apakah kode promo valid? (y/t): ").strip().lower()
kode_promo_valid_2010 = input_promo_2010 in ["y", "t"]

total_diskon_persen_2010 = 0

# Multi-if terpisah: Setiap kondisi diperiksa secara independen
# Diskon bisa ditumpuk (akumulasi) jika memenuhi beberapa syarat sekaligus

if total_belanja_2010 > 1000000:
    total_diskon_persen_2010 += 10  # Diskon total besar

if is_member_2010:
    total_diskon_persen_2010 += 5  # Diskon member

if kode_promo_valid_2010:
    total_diskon_persen_2010 += 15  # Diskon voucher

# Menghitung nominal diskon dan total bayar
nominal_diskon_2010 = total_belanja_2010 * (total_diskon_persen_2010 / 100)
total_bayar_2010 = total_belanja_2010 - nominal_diskon_2010

# Output hasil
print("\n--- Rincian Pembayaran ---")
print(f"Total Diskon  : {total_diskon_persen_2010} (Rp {nominal_diskon_2010:.0f})")
print(f"Total Bayar   : Rp {total_bayar_2010:.0f}")

print(f"Total diskon yang Anda dapatkan: {total_diskon_persen_2010}%")
# Output: Total diskon yang Anda dapatkan: 30% jika belanja > 1 juta, member, dan kode promo valid
```

<div >

Penjelasan:
1. Program dimulai dengan meminta pengguna memasukkan **total belanja** dalam rupiah menggunakan fungsi `input()`.
2. Nilai total belanja diubah menjadi tipe data `float` dan disimpan dalam variabel `total_belanja_2010`.
3. Program meminta pengguna memasukkan status member dengan pilihan `y` atau `t`. Input kemudian diubah menjadi huruf kecil menggunakan `.lower()` dan spasi yang tidak diperlukan dihapus menggunakan `.strip()`.
4. Program memeriksa apakah input member berupa `"y"` atau `"t"` menggunakan operator `in`. Hasil pemeriksaan disimpan dalam variabel `is_member_2010`.
5. Program meminta pengguna memasukkan status kode promo dengan pilihan `y` atau `t`. Input diproses dengan cara yang sama dan hasilnya disimpan dalam variabel `kode_promo_valid_2010`.
6. Variabel `total_diskon_persen_2010` diberikan nilai awal `0` sebagai tempat untuk menyimpan jumlah persentase diskon.
7. Program memeriksa kondisi pertama. Jika total belanja lebih dari **Rp1.000.000**, program menambahkan diskon sebesar **10%**.
8. Program memeriksa kondisi kedua. Jika pengguna merupakan member, program menambahkan diskon sebesar **5%**.
9. Program memeriksa kondisi ketiga. Jika kode promo dianggap valid, program menambahkan diskon sebesar **15%**.
10. Karena menggunakan beberapa `if` yang terpisah, setiap kondisi diperiksa secara independen. Jika beberapa kondisi terpenuhi, persentase diskon akan **diakumulasikan**.
11. Program menghitung nominal diskon dengan mengalikan total belanja dengan persentase diskon yang telah diperoleh.
12. Program menghitung total pembayaran dengan mengurangi total belanja dengan nominal diskon.
13. Program menampilkan rincian pembayaran berupa total persentase diskon, nominal diskon dalam rupiah, dan total yang harus dibayar.
14. Program menampilkan kembali jumlah diskon yang diperoleh dalam bentuk persentase.

### 2.2.7 Program match_case1_2611532010
</div>

```python
# Buat file dengan nama case1.py
# Buat program untuk kodisional if
# Nama variabel ditambah 4 digit terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input()
# Program konversi angka menjadi nama bulan

bulan_2010 = int(input("Masukkan angka bulan (1 - 12): "))

match bulan_2010:
    case 1:
        print("Januari")
    case 2:
        print("Februari")
    case 3:
        print("Maret")
    case 4:
        print("April")
    case 5:
        print("Mei")
    case 6:
        print("Juni")
    case 7:
        print("Juli")
    case 8:
        print("Agustus")
    case 9:
        print("September")
    case 10:
        print("Oktober")
    case 11:
        print("November")
    case 12:
        print("Desember")
    case _:
        print("Angka tidak valid")
```

<div >

Penjelasan:
1. Program dimulai dengan meminta pengguna memasukkan angka bulan dari **1 sampai 12** menggunakan fungsi `input()`.
2. Nilai yang dimasukkan diubah menjadi tipe data `int` dan disimpan dalam variabel `bulan_2010`.
3. Program menggunakan struktur **`match-case`** untuk mencocokkan nilai `bulan_2010` dengan beberapa pilihan yang tersedia.
4. Jika nilai yang dimasukkan adalah **1**, program menampilkan **Januari**.
5. Jika nilai **2 sampai 12**, program menampilkan nama bulan yang sesuai, mulai dari Februari hingga Desember.
6. Jika angka yang dimasukkan tidak sesuai dengan angka **1 sampai 12**, program menjalankan `case _` dan menampilkan pesan **"Angka tidak valid"**.
7. Setelah nama bulan atau pesan kesalahan ditampilkan, program selesai.

```python
# === SISTEM LOKET ASTRA ADVENTURE PARK ===

# 1. Input Data Pengunjung & String Handling
nama_pengunjung_2010 = input("Masukkan Nama Pengunjung        : ")
umur_2010 = int(input("Input umur anda                  : "))
sim_2010 = input("Apakah Anda Sudah Punya SIM C (y/t): ").strip().lower()

# Menu Paket Wahana di Astra Adventure Park
print("\nPilihan Paket Wahana (1-5):")
print("  1. Safari Rimba         (Rp 50,000)")
print("  2. Arung Jeram          (Rp 75,000)")
print("  3. Motor ATV Ekstrim    (Rp 120,000)")
print("  4. Roller Coaster Kilat (Rp 100,000)")
print("  5. All-Access VIP       (Rp 220,000)")

paket_2010 = input("Masukkan nomor paket (1-5)      : ")

# 2. Pemilihan Wahana Menggunakan match-case
match paket_2010:
    case "1":
        nama_paket_2010 = "Wahana Safari Rimba"
        harga_satuan_2010 = 50000
    case "2":
        nama_paket_2010 = "Wahana Arung Jeram"
        harga_satuan_2010 = 75000
    case "3":
        nama_paket_2010 = "Wahana Motor ATV Ekstrim"
        harga_satuan_2010 = 120000
    case "4":
        nama_paket_2010 = "Wahana Roller Coaster Kilat"
        harga_satuan_2010 = 100000
    case "5":
        nama_paket_2010 = "Wahana All-Access VIP"
        harga_satuan_2010 = 220000
    case _:
        print("Paket wahana tidak valid!")
        exit()

jumlah_tiket_2010 = int(input("Masukkan jumlah tiket            : "))

# Validation: IF Tunggal
if jumlah_tiket_2010 <= 0:
    print("Peringatan: Kuota tiket tidak valid!")
    exit()

is_member_2010 = input("Apakah Anda member? (y/t)        : ").strip().lower()
kode_promo_valid_2010 = input("Apakah kode promo valid? (y/t)  : ").strip().lower()

# 3. Validasi Izin Kendali Wahana Menggunakan if-elif-else / if-else
print("\n--- KELAYAKAN PENGENDARA WAHANA ---")
if paket_2010 == "3":
    if umur_2010 >= 17 and sim_2010 in ['y', 'ya']:
        print("Status Akses: Anda sudah dewasa dan boleh mengendarai ATV sendiri.")
    elif umur_2010 >= 17 and sim_2010 not in ['y', 'ya']:
        print("Status Akses: Anda sudah dewasa tetapi tidak boleh bawa motor ATV (wajib didampingi instruktur).")
    elif umur_2010 < 17 and sim_2010 in ['y', 'ya']:
        print("Status Akses: Identitas tidak valid: Belum cukup umur memiliki SIM.")
    else:
        print("Status Akses: Anda belum cukup umur dan tidak boleh bawa motor ATV.")
else:
    if umur_2010 >= 10:
        print("Status Akses: Pengunjung memenuhi batas usia minimum wahana.")
    else:
        print("Status Akses: Pengunjung belum memenuhi batas usia minimum wahana.")

# 4. Akumulasi Diskon Bertingkat Menggunakan Multi-IF Terpisah
subtotal_2010 = harga_satuan_2010 * jumlah_tiket_2010
total_diskon_persen_2010 = 0

if subtotal_2010 >= 200000:
    total_diskon_persen_2010 += 10

if is_member_2010 in ['y', 'ya']:
    total_diskon_persen_2010 += 5

if kode_promo_valid_2010 in ['y', 'ya']:
    total_diskon_persen_2010 += 15

if jumlah_tiket_2010 >= 5:
    total_diskon_persen_2010 += 5

# 5. Perhitungan & Evaluasi Audit Menggunakan if-else
nominal_diskon_2010 = subtotal_2010 * (total_diskon_persen_2010 / 100)
total_bayar_2010 = subtotal_2010 - nominal_diskon_2010

if total_bayar_2010 > 300000:
    catatan_layanan_2010 = "Selamat! Anda berhak mendapatkan Souvenir Gratis."
else:
    catatan_layanan_2010 = "Terima kasih telah berkunjung."

# Tampilan Rincian Pembayaran
print("\n--- Rincian Pembayaran ---")
print(f"Nama Pengunjung  : {nama_pengunjung_2010}")
print(f"Umur             : {umur_2010} tahun")
print(f"Subtotal Belanja : Rp {subtotal_2010:,.0f}")
print(f"Total Diskon     : {total_diskon_persen_2010}% (Rp {nominal_diskon_2010:,.0f})")
print(f"Total Bayar      : Rp {total_bayar_2010:,.0f}")
print(f"Catatan Layanan  : {catatan_layanan_2010}")
print("Program Selesai")
```
