## 2.1 Dasar Teori
1. **Perulangan `for`:** Perulangan ini digunakan untuk mengiterasi suatu barisan (*sequence* seperti *list, tuple, string*) atau dieksekusi sejumlah iterasi numerik yang dipatok menggunakan fungsi tertentu. Berbeda dengan *while-loop* yang bergantung murni pada kondisi Boolean *(True/False)*, *for-loop* memiliki ujung iterasi yang jauh lebih terprediksi.
2. **Fungsi `range()`:** Sering mendampingi `for`, fungsi ini menghasilkan deret bilangan berurutan. Format utamanya adalah `range(start, stop, step)`. Nilai *stop* bersifat eksklusif (tidak diikutsertakan dalam eksekusi) sehingga untuk mencapai iterasi `N`, argumen yang diberikan idealnya adalah `N+1`.
3. **Nested Loop (Perulangan Bersarang):** Merupakan skenario peletakan blok perulangan di dalam kurung perulangan yang lain. Variabel perulangan di bagian luar (*outer loop*) umumnya mengendalikan pergantian dan lompatan baris (vertikal), sedangkan perulangan di sisi dalam (*inner loop*) mengendalikan perulangan karakter per karakter pada struktur baris tersebut secara memanjang ke samping (horizontal).

## 2.2 Kode Program

### 2.2.1 Program "perulangan_for1_2611532010.py"


```python
# BUat file dengan nama perulangan_for1_nim.py
# Buat program untuk perulangan for dalam python
# Nama variabel ditambah 4 digit terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

ulang_2010 = int(input("Masukkan jumlah perulangan: "))

for i_2010 in range(ulang_2010):
    print(f"Perulangan ke-{i_2010}") 
```

**Penjelasan :**
Program ini menunjukkan konsep paling dasar dari perulangan `for`. Dengan menggunakan fungsi `range(ulang)`, program melakukan iterasi mulai dari indeks `0` hingga batas `ulang - 1`. Pada setiap iterasinya, program mencetak string "Perulangan ke-" yang digabungkan dengan nilai indeks iterasi saat itu.

### 2.2.2 Program "perulangan_for2_2611532010.py"


```python
# Buat file dengan nama perulangan_for2_nim.py
# Buat program untuk perulangan for dalam python
# Nama variabel ditambah 4 digit terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

ulang_2010 = int(input("Masukkan jumlah perulangan: "))
print("Perulangan ke-0 sampai ke-", ulang_2010-1)
for i_2010 in range(ulang_2010):
    print(i_2010, end=" ")
print()
print("perulangan ke-1 sampai ke-", ulang_2010)
for i_2010 in range(1, ulang_2010+1):
    print(i_2010, end=" ")
```

**Penjelasan:**

Program ini mendemonstrasikan cara memanipulasi parameter batas awal dan batas akhir pada fungsi `range()`. Perulangan pertama menggunakan `range(ulang)` yang secara bawaan dimulai dari angka 0. Perulangan kedua memodifikasi rentang menjadi `range(1, ulang + 1)` agar iterasi dimulai dari angka 1 dan berakhir tepat pada nilai input pengguna. Parameter `end=" "` ditambahkan agar hasil cetakan tersusun menyamping (horizontal).

### 2.2.3 Program "perulangan_for3_2611532010.py"


```python
# Buat file dengan nama perulangan_for3_nim.py
# Buat program untuk perulangan for dalam python
# Nama variabel ditambah 4 digit terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

ulang_2010 = int(input("Masukkan jumlah perulangan: "))

jumlah_2010 = 0
for i_2010 in range(1, ulang_2010 + 1):
    print(i_2010, end=" ")
    jumlah_2010 = jumlah_2010 + i_2010

    if i_2010 < ulang_2010:
        print("+", end=" ")
    else:
        print(" = ", jumlah_2010, end=" ")
print()
print("jumlah_2010 =", jumlah_2010)
```

**Penjelasan:**

Program ini mengimplementasikan logika akumulasi dengan bantuan variabel penyimpan `jumlah`. Pada setiap perulangan, angka yang sedang diiterasi akan dicetak dan ditambahkan ke dalam variabel `jumlah`. Selain itu, program memanfaatkan percabangan `if-else` untuk menentukan kapan harus mencetak simbol `+` (jika iterasi masih berlangsung) dan kapan harus mencetak simbol `=` beserta hasil totalnya (jika iterasi sudah mencapai angka terakhir).

### 2.2.4 Program "jumlah_genap_2611532010.py"


```python
# Buat file dengan nama jumlah_genap_nim.py
# Buat program untuk menghitung jumlah bilangan genap
# Nama variabel ditambah 4 digit terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

ulang_2010 = int(input("masukkan nilai batas: "))

jumlah_genap_2010 = 0
for i_2010 in range(1, ulang_2010 + 1):
    if i_2010 % 2 == 0:
        print(i_2010, end=" ")
        jumlah_genap_2010 += i_2010

        if i_2010 < ulang_2010:
            print("+", end=" ")
        else:
            print(" = ", jumlah_genap_2010, end="")
print()
print("jumlah =", jumlah_genap_2010)
```

**Penjelasan :**
Program ini merupakan pengembangan dari logika akumulasi sebelumnya dengan penambahan filter kondisional. Terdapat struktur `if i % 2 == 0:` di dalam perulangan yang berfungsi untuk mengecek sisa bagi. Dengan kondisi ini, program hanya akan mencetak, memproses, dan menjumlahkan angka-angka yang bernilai genap saja, sedangkan angka ganjil akan dilewati (*skip*).

### 2.2.5 Program "nested_for1_2611532010.py"


```python
# Buat file dengan nama nested_for1_nim.py
# Buat program untuk perulangan for dalam python
# Nama variabel ditambah 4 digit terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

batas_2010 = int(input("masukkan nilai batas:  "))
for line in range(1, batas_2010 + 1):
    for j_2010 in range(1, (-1 * line + batas_2010 + 1)):
        print(".", end="")
    print(line)
```

**Penjelasan :**
Program ini memperkenalkan perulangan bersarang (*nested loop*). *Outer loop* (perulangan luar) bertugas mengatur perpindahan baris, sedangkan *inner loop* (perulangan dalam) bertugas mencetak karakter titik (`.`). Jumlah karakter titik yang dicetak bersifat dinamis dan semakin menurun pada setiap barisnya, dikendalikan oleh formula matematika `(-1 * line + batas)`. Pada bagian ujung tiap baris, nilai baris tersebut dicetak.

### 2.2.6 Program "nested_for2_2611532010.py"


```python
# Buat file dengan nama nested_for2_nim.py
# Buat program untuk perulangan for dalam python
# Nama variabel ditambah 4 digit terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

batas_2010 = int(input("masukkan nilai batas: "))
for i_2010 in range(1, batas_2010 + 1):
    for j_2010 in range(1, batas_2010 + 1):
        print("*", end="")
    print() # pindah ke baris berikutnya
```

**Penjelasan :**
Program ini membentuk pola bangun datar persegi menggunakan karakter asterisk (`*`). *Outer loop* melakukan iterasi untuk menentukan jumlah baris ke bawah, sementara *inner loop* melakukan iterasi untuk mencetak karakter bintang pada setiap kolom menyamping. Sebuah perintah `print()` kosong diletakkan di akhir *outer loop* sebagai pemicu (*trigger*) untuk pindah ke baris baru.

### 2.2.7 Program "nested_for3_2611532010.py"


```python
# Buat file dengan nama nested_for3_nim.py
# Buat program untuk perulangan for dalam python
# Nama variabel ditambah 4 digit terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

batas_2010 = int(input("masukkan nilai batas:"))
for i_2010 in range(batas_2010+1):
    for j_2010 in range(batas_2010+1):
        print (i_2010+j_2010, end=" ")
    print() # pindah ke baris berikutnya
```

**Penjelasan :**
Program ini menampilkan matriks numerik dua dimensi yang nilainya terus meningkat. Logikanya berpusat pada relasi antara indeks baris (`i`) dan indeks kolom (`j`). Di dalam *inner loop*, program mencetak hasil penjumlahan dari posisi baris dan posisi kolom (`i + j`). Hal ini menghasilkan deret angka yang membentuk pola diagonal yang berurutan.

### 2.2.8 Program "nested_for4_2611532010.py"


```python
# Buat file dengan nama nested_for4_nim.py
# Buat program untuk perulangan for dalam python
# Nama variabel ditambah 4 digit terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

tinggi_2010 = int(input("masukkan tinggi pola (bilangan genap, misal 10):"))

if tinggi_2010 % 2 != 0:
    print("Tinggi harus bilangan genap!")
else:
    a_2010 = tinggi_2010 
    c_2010 = a_2010
    lebar_2010 = (2 * tinggi_2010)- 2

    for i_2010 in range(1, tinggi_2010 +1):
        b_2010 = c_2010 + 1

        for j_2010 in range (1, lebar_2010 + 1):

            # Baris atas dan bawah
            if i_2010 == 1 or i_2010 == tinggi_2010:
                if j_2010 == 1 or j_2010 == lebar_2010:
                    print("#", end="")
                else:
                    print("=", end="")

                # Baris isi
            else:
                if j_2010 == 1 or j_2010 == lebar_2010:
                    print("|", end="")
                else:
                    if j_2010 == c_2010:
                        print("<", end="")
                    elif j_2010 == b_2010:
                        print(">", end="")
                    elif j_2010 == (lebar_2010 - c_2010):
                        print("<", end="")
                    elif j_2010 == (lebar_2010 - c_2010 + 1):
                        print(">", end="")
                    elif j_2010 > b_2010 and j_2010 < (lebar_2010 - c_2010):
                        print(".", end="")
                    else:
                        print(" ", end="")
        
        print()

        # logika asli java
        a_2010 -= 2

        if a_2010 <= 0:
            c_2010 = (-a_2010) + 2
        else:
            c_2010 = a_2010
```

**Penjelasan :**
Program ini menggunakan *nested loop* kompleks untuk membentuk sebuah bangun simetris dua dimensi yang dilengkapi dengan bingkai luar. *Outer loop* menangani tinggi pola, sementara *inner loop* menyusun setiap piksel karakter secara horizontal. Program ini sangat bergantung pada struktur percabangan tingkat lanjut (`if-elif-else`) dan variabel kalkulasi geometri (`a`, `b`, `c`, `lebar`) untuk menentukan letak koordinat yang pas guna mencetak batas pagar (`#`, `=`), dinding (`|`), maupun corak bagian dalam (`<`, `>`, `.`, atau spasi kosong).

### 2.2.9 Program "latihan5_[NIM].py"


```python
# Latihan 5 membuat program untuk perulangan for (segitiga bintang)

tinggi_2010 = int(input("Masukkan tinggi segitiga: "))

for i_2010 in range(1, tinggi_2010 + 1):
    for j_2010 in range(tinggi_2010 - i_2010):
        print(" ", end="")
    for k_2010 in range(i_2010):
        print("* ", end="")
    print()
```

**Penjelasan :**
Program ini menggunakan perulangan bersarang untuk mencetak pola segitiga sama kaki (piramida) menggunakan karakter bintang (`*`). *Outer loop* bertugas mengendalikan jumlah baris dari atas ke bawah sesuai dengan tinggi yang diinputkan pengguna. Di dalam *outer loop* tersebut, terdapat dua *inner loop* yang berjalan secara berurutan: *inner loop* pertama mencetak spasi kosong yang jumlahnya semakin menurun di tiap baris agar pola terdorong ke tengah, sedangkan *inner loop* kedua mencetak karakter bintang yang diikuti satu spasi (`* `) dengan jumlah yang semakin meningkat. Perpaduan spasi penyeimbang dan penambahan karakter bintang ini menghasilkan bentuk bangun piramida yang simetris dan rapi.

## Tugas Praktikum

**Soal :**

Studi Kasus: Pola Jam Pasir Kristal Palindromik Berbingkai (Framed Palindromic Crystal Hourglass)
Mahasiswa diminta merancang program konsol interaktif untuk mencetak pola bangun geometri dua dimensi berupa Jam Pasir Kristal Palindromik yang terbungkus rapi di dalam bingkai pembatas berpagar. Berbeda dengan pola dasar, program ini menuntut pembentukan deret angka yang bergerak dinamis menurun dan menaik secara palindromik pada setiap baris serta diselaraskan dengan poros kristal di pusat bangun.
Program menerima masukan nilai integer positif N dari pengguna yang merepresentasikan ukuran skala jam pasir (skala angka tertinggi pada puncak dan alas).

**Kode Jawaban :**

a. Kode Program "tugas5_2611532010.py"

```python
# === PROGRAM JAM PASIR KRISTAL PALINDROMIK ===

# Input ukuran skala jam pasir N
n_2010 = int(input("Masukkan ukuran skala jam pasir (N): "))

# 1. BORDER ATAS
print("#", end="")
for i_2010 in range(4 * n_2010 + 5):
    print("=", end="")
print("#")

# 2. FASE 1: JAM PASIR ATAS (Baris N turun ke 1)
for baris_2010 in range(n_2010, 0, -1):
    # Dinding kiri
    print("| ", end="")
    
    # Spasi penyeimbang kiri: 2 * (N - baris)
    for spasi_2010 in range(2 * (n_2010 - baris_2010)):
        print(" ", end="")
        
    # Deret angka mundur dari baris ke 1
    for angka_2010 in range(baris_2010, 0, -1):
        print(angka_2010, end=" ")
        
    # Poros kristal tengah
    print("<*>", end="")
    
    # Deret angka maju dari 1 ke baris
    for angka_2010 in range(1, baris_2010 + 1):
        print("", angka_2010, end="")
        
    # Spasi penyeimbang kanan: 2 * (N - baris)
    for spasi_2010 in range(2 * (n_2010 - baris_2010)):
        print(" ", end="")
        
    # Dinding kanan
    print(" |")

# 3. FASE 2: POROS TITIK PUSAT (SINGULARITY)
print("|", end="")

# Spasi penyeimbang kiri: 2 * N + 1
for spasi_2010 in range(2 * n_2010 + 1):
    print(" ", end="")

# Poros kristal tunggal
print("<*>", end="")

# Spasi penyeimbang kanan: 2 * N + 1
for spasi_2010 in range(2 * n_2010 + 1):
    print(" ", end="")

print("|")

# 4. FASE 3: JAM PASIR BAWAH (Baris 1 naik ke N)
for baris_2010 in range(1, n_2010 + 1):
    # Dinding kiri
    print("| ", end="")
    
    # Spasi penyeimbang kiri: 2 * (N - baris)
    for spasi_2010 in range(2 * (n_2010 - baris_2010)):
        print(" ", end="")
        
    # Deret angka mundur dari baris ke 1
    for angka_2010 in range(baris_2010, 0, -1):
        print(angka_2010, end=" ")
        
    # Poros kristal tengah
    print("<*>", end="")
    
    # Deret angka maju dari 1 ke baris
    for angka_2010 in range(1, baris_2010 + 1):
        print("", angka_2010, end="")
        
    # Spasi penyeimbang kanan: 2 * (N - baris)
    for spasi_2010 in range(2 * (n_2010 - baris_2010)):
        print(" ", end="")
        
    # Dinding kanan
    print(" |")

# 5. BORDER BAWAH
print("#", end="")
for i_2010 in range(4 * n_2010 + 5):
    print("=", end="")
print("#")
```

**Penjelasan :**

Program ini merupakan implementasi *nested loop* (perulangan bersarang) tingkat lanjut untuk mencetak bangun geometri dua dimensi secara dinamis tanpa menggunakan trik manipulasi string (seperti perkalian karakter). Logika program ini dibagi menjadi beberapa fase eksekusi:

1. **Bagian pertama (Input Ukuran Skala)** berfungsi untuk menerima input dari pengguna berupa nilai `n_2010` yang menentukan ukuran atau skala dari pola jam pasir yang akan dibuat. Nilai tersebut diubah menjadi tipe data integer menggunakan `int(input())` agar dapat digunakan dalam proses perulangan.

2. **Bagian kedua (Border Atas)** berfungsi untuk mencetak bingkai bagian atas jam pasir dengan pola `#====#`. Panjang tanda `=` dibuat secara dinamis menggunakan perulangan `for` dengan jumlah `4 * n_2010 + 5`, sehingga panjang border akan menyesuaikan dengan nilai `n_2010` yang dimasukkan pengguna.

3. **Bagian ketiga (Fase 1 - Jam Pasir Atas)** berfungsi untuk mencetak pola jam pasir bagian atas menggunakan perulangan `baris_2010` dari nilai `N` turun hingga `1`. Pada setiap baris, program mencetak dinding kiri `|`, spasi penyeimbang, deret angka yang tersusun mundur dari nilai baris menuju `1`, simbol poros kristal `<*>`, kemudian deret angka yang tersusun maju dari `1` kembali menuju nilai baris. Spasi di sisi kiri dan kanan digunakan untuk membentuk pola jam pasir agar semakin menyempit menuju bagian tengah.

4. **Bagian keempat (Fase 2 - Poros Titik Pusat / Singularity)** berfungsi untuk mencetak bagian tengah atau titik pusat jam pasir. Bagian ini terdiri dari dinding kiri `|`, spasi penyeimbang di kedua sisi, dan satu simbol poros kristal `<*>` sebagai pusat dari pola jam pasir.

5. **Bagian kelima (Fase 3 - Jam Pasir Bawah)** berfungsi untuk mencetak pola jam pasir bagian bawah menggunakan perulangan `baris_2010` dari `1` naik hingga `N`. Struktur angka dan spasi dibuat sama seperti bagian atas, tetapi urutannya dibalik sehingga membentuk pola yang simetris dengan Fase 1. Deret angka tetap tersusun secara palindromik dengan simbol `<*>` sebagai poros di tengah.

6. **Bagian keenam (Border Bawah)** berfungsi untuk mencetak bingkai penutup bagian bawah jam pasir. Bentuknya sama dengan border atas, yaitu diawali dan diakhiri dengan simbol `#`, sedangkan bagian tengahnya terdiri dari tanda `=` yang jumlahnya menyesuaikan dengan nilai `n_2010`.

Secara keseluruhan, program ini menggunakan **perulangan `for`**, **perhitungan jumlah spasi secara dinamis**, serta **pola angka palindromik** untuk membentuk sebuah jam pasir yang memiliki poros kristal `<*>` di bagian tengah. Ukuran pola dapat berubah sesuai dengan nilai `N` yang dimasukkan oleh pengguna.
